```python
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf
```

```python
xs = np.arange(-1,5, dtype = float)
ys = np.array([-4.0, 1.0, 6.0, 11.0, 16.0, 21.0])

print(f'xs{xs}')
print(f'ys{ys}')
```

```python
regresion_lineal = np.polyfit(xs, ys, 1)
beta, alpha = regresion_lineal
print(f'beta: {beta}')
print(f'alpha: {alpha}')
print(f'y={round(beta, 1)}x + {round(alpha, 1)}')
```

```python
plt.plot(xs, ys, 'o', color = 'green', label = 'Datos')
plt.plot(xs, beta*xs + alpha, color = 'blue', label = 'regresion lineal')
plt.axhline(y = 0, color = 'red')
plt.axvline(x = 0, color = 'red')
plt.title('Say hello to the "Hello, World" of ML')
plt.xlabel('xs')
plt.ylabel('ys')
plt.legend()
plt.grid()
plt.show()
```

```python
capa = tf.keras.layers.Dense(units = 1)
modelo = tf.keras.Sequential([
    tf.keras.Input(shape = (1,)),
    capa
])

modelo.compile(optimizer='sgd', loss = 'mse')
```

```python
modelo.summary()
```

```python
print('Comenzando entrenamiento...')
historial = modelo.fit(xs, ys, epochs = 500, verbose = True)
print('modelo entrenado!')
```

```python
valor = float(input('Ingrese un valor de x para predecir y: '))
prediccion = modelo.predict(np.array([[valor]]))
print(f'El valor de y para x = {valor} es: {prediccion}')
```

```python
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import classification_report, confusion_matrix


X = xs.reshape(-1, 1)

#conversion de datos.
bins = [-np.inf, 0, 50, np.inf]
y_categorical = np.digitize(ys, bins) - 1 # np.digitize devuelve el índice del bin comenzando desde 1, por lo que se resta 1 para obtener clases indexadas desde 0.

print("Original features (X):")
print(X.flatten())
print("\nOriginal labels (ys):")
print(ys)
print("\nCategorical labels (y) for classification:")
print(y_categorical)

# Entrenamiento de los datos.
X_train, X_test, y_train, y_test = train_test_split(X, y_categorical, test_size=0.3, random_state=42)

print("\n--- Data Split ---")
print(f"X_train shape: {X_train.shape}, y_train shape: {y_train.shape}")
print(f"X_test shape: {X_test.shape}, y_test shape: {y_test.shape}")

# 2. Normalizamos datos
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

print("\n--- Data Scaling ---")
print("X_train_scaled (first 5 rows):\n", X_train_scaled[:5])
print("X_test_scaled (first 5 rows):\n", X_test_scaled[:5])

# 3. Entrenamos el KNN
knn_classifier = KNeighborsClassifier(n_neighbors=3)
knn_classifier.fit(X_train_scaled, y_train)

print("\n--- KNN Model Training Complete ---")

# 4. Evaluamos el clasificador KNN
y_pred = knn_classifier.predict(X_test_scaled)

print("\n--- Model Evaluation ---")
print("Classification Report:")
print(classification_report(y_test, y_pred))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred))
```

```python
from sklearn.metrics import accuracy_score
import matplotlib.pyplot as plt



# Definir un rango de valores de k a probar.
# El valor máximo de k debe ser <= len(X_train_scaled).
# Dado que X_train_scaled tiene 4 muestras, k puede ser 1, 2, 3 o 4.

# En KNN, normalmente k suele ser un número impar para evitar empates,
# pero con datasets muy pequeños también podemos probar números pares
# y observar el comportamiento del modelo.
k_values = range(1, len(X_train_scaled) + 1)
errors = []
accuracies = []

print("--- Hyperparameter Tuning for K in KNN ---")
print(f"Testing k values from {min(k_values)} to {max(k_values)}")

for k in k_values:
    # Train KNN model for current k
    knn = KNeighborsClassifier(n_neighbors=k)
    knn.fit(X_train_scaled, y_train)

    # Predict on the test set
    y_pred_k = knn.predict(X_test_scaled)

    # Calculate accuracy and error
    accuracy = accuracy_score(y_test, y_pred_k)
    error = 1 - accuracy

    errors.append(error)
    accuracies.append(accuracy)
    print(f"k = {k}: Accuracy = {accuracy:.2f}, Error = {error:.2f}")

# Visualize the results
plt.figure(figsize=(8, 5))
plt.plot(k_values, errors, marker='o', linestyle='-')
plt.title('KNN Error Rate vs. K Value')
plt.xlabel('Number of Neighbors (K)')
plt.ylabel('Error Rate (1 - Accuracy)')
plt.xticks(k_values)
plt.grid(True)
plt.show()

# Find the optimal k that minimizes the error
optimal_k_index = np.argmin(errors)
optimal_k = k_values[optimal_k_index]
min_error = errors[optimal_k_index]

print(f"\nOptimal K value that minimizes error: {optimal_k}")
print(f"Minimum Error Rate: {min_error:.2f}")
```
